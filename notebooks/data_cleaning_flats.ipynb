```python
df.drop(columns='link',inplace=True)
```

```python
#location
df['location'].value_counts()
```

```python
df['location']=df['location'].str.split(',').str.get(0).str.lower()
```

```python
#area
df['area']=df['area'].str.split().str.get(0).str.replace(',', '').astype(float)*9
```

```python
df.head()
```

```python
#price
def convert_price(price):
    if pd.isna(price):
        return None

    price = str(price).replace(",", "").strip().lower()

    if "arab" in price:
        return float(price.replace("arab", "").strip()) * 1_000_000_000

    elif "crore" in price:
        return float(price.replace("crore", "").strip()) * 10_000_000

    elif "lakh" in price:
        return float(price.replace("lakh", "").strip()) * 100_000

    else:
        return float(price)

df['price']=df['price'].apply(convert_price)
```
