```python
import re

def format_sub_location(x):
    if pd.isna(x):
        return x

    x = str(x).strip().lower()

    # This section handles Remove Extra Spaces.
    x = re.sub(r"\s+", " ", x)

    # Remove parent names from sub_location
    x = re.sub(r"^north nazimabad\s*-\s*", "", x)
    x = re.sub(r"^federal b area\s*", "", x)
    x = re.sub(r"^dha\s*", "", x)
    x = re.sub(r"^scheme 33\s*", "", x)

    # This section handles Block Formatting.
    x = re.sub(r"block\s*([a-z0-9]+)", lambda m: f"block {m.group(1)}", x)

    # This section handles Phase Formatting.
    x = re.sub(r"phase\s*(\d+[a-z]?)", lambda m: f"phase {m.group(1)}", x)

    # This section handles Sector Formatting.
    x = re.sub(r"sector\s*(\d+[a-z]?-\w?)", lambda m: f"sector {m.group(1)}", x)

    # This section handles Title Case.
    x = x.title()

    return x


df["sub_location"] = df["sub_location"].apply(format_sub_location)
```

```python
df["luxury_score"] = (
    (df["parking"] > 0).astype(int) +
    (df["servant_quarters"] > 0).astype(int) +
    df["elevator"].astype(int) +
    df["electricity_backup"].astype(int) +
    df["furnished"].astype(int)
)
```

```python
# 1. Convert all potential count features to binary flags (1 or 0)
parking_bin = (df["parking"] > 0).astype(int)
servant_bin = (df["servant_quarters"] > 0).astype(int)
elevator_bin = (df["elevator"] > 0).astype(int)
elec_backup_bin = (df["electricity_backup"] > 0).astype(int)
furnished_bin = (df["furnished"] > 0).astype(int)

# 2. Initialize luxury_score column
df["luxury_score"] = 0

# 3. Create masks
flat_mask = df["property_type"].str.lower() == "flat"
house_mask = df["property_type"].str.lower() == "house"

# 4. Calculate for Flats
df.loc[flat_mask, "luxury_score"] = (
    1 * parking_bin[flat_mask] +
    2 * elevator_bin[flat_mask] +
    2 * elec_backup_bin[flat_mask] +
    2 * servant_bin[flat_mask] +
    1 * furnished_bin[flat_mask]
)

# 5. Calculate for Houses
df.loc[house_mask, "luxury_score"] = (
    2 * parking_bin[house_mask] +
    2 * elec_backup_bin[house_mask] +
    3 * servant_bin[house_mask] +
    1 * furnished_bin[house_mask]
)
```

```python
import pandas as pd

# Step 1: Median price/sqft for each sub-location
sub_stats = (
    df.groupby("sub_location")
      .agg(
          median_ppsqft=("price_per_sqft", "median"),
          count=("price_per_sqft", "size")
      )
      .reset_index()
)

# Step 2: Remove effect of very small groups
MIN_COUNT = 10

valid = sub_stats["count"] >= MIN_COUNT

sub_stats.loc[valid, "global_percentile"] = (
    sub_stats.loc[valid, "median_ppsqft"]
             .rank(method="average", pct=True)
)

# This section handles Fill Rare Sub-Locations With Median Percentile.
sub_stats["global_percentile"] = (
    sub_stats["global_percentile"]
    .fillna(sub_stats.loc[valid, "global_percentile"].median())
)

# Step 3: Merge back
df = df.merge(
    sub_stats[["sub_location", "global_percentile"]],
    on="sub_location",
    how="left"
)

# Step 4: Normalize parent location score
# (1-5 --> 0-1)
parent_norm = (df["location_score"] - 1) / 4

# Step 5: Combine
combined_score = (
    0.8 * df["global_percentile"] +
    0.2 * parent_norm
)

# Step 6: Convert to 1-5 integer score
df["sub_location_score"] = pd.qcut(
    combined_score.rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(float)
```

```python
import pandas as pd

# Step 1: Calculate median price/sqft for each sub-location
sub_stats = (
    df.groupby("sub_location")
      .agg(
          median_ppsqft=("price_per_sqft", "median"),
          count=("price_per_sqft", "size")
      )
      .reset_index()
)


# Step 2: Calculate global percentile of sub-locations
MIN_COUNT = 10

valid = sub_stats["count"] >= MIN_COUNT

sub_stats.loc[valid, "global_percentile"] = (
    sub_stats.loc[valid, "median_ppsqft"]
    .rank(method="average", pct=True)
)

# This section handles Handle Rare Sub-Locations.
median_pct = sub_stats.loc[valid, "global_percentile"].median()

sub_stats["global_percentile"] = (
    sub_stats["global_percentile"]
    .fillna(median_pct)
)


# Step 3: Map percentile back to main dataframe
pct_map = sub_stats.set_index("sub_location")["global_percentile"]

df["global_percentile"] = df["sub_location"].map(pct_map)


# Step 4: Normalize parent location score
# 1-5 becomes 0-1
parent_norm = (df["location_score"] - 1) / 4


# Step 5: Combine parent score + market value
combined_score = (
    0.8 * df["global_percentile"] +
    0.2 * parent_norm
)


# Step 6: Convert into integer score 1-5
percentile = combined_score.rank(pct=True)

df["sub_location_score"] = pd.cut(
    percentile,
    bins=[0, 0.2, 0.4, 0.6, 0.8, 1],
    labels=[1, 2, 3, 4, 5],
    include_lowest=True
).astype(float)


# Step 7: Remove helper columns (optional)
df.drop(
    columns=["global_percentile"],
    inplace=True
)
```
