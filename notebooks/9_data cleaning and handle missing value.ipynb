```python
import re
import pandas as pd

SECTOR_RE = re.compile(
    r'\bsector\.?\s*-?\s*([0-9]+[a-z0-9]*(?:[\-/][a-z0-9]+)*)(?:\s+([a-z])\b)?',
    re.IGNORECASE
)

def extract_sector(text):
    if not isinstance(text, str):
        return None
    m = SECTOR_RE.search(text)
    if m:
        val = m.group(1).upper()
        if m.group(2):
            val += m.group(2).upper()
        return f"Sector {val}"
    return None

extracted = x.apply(extract_sector)
extracted = extracted.apply(lambda v: f"North Karachi {v}" if pd.notna(v) else None)


df.loc[extracted.index, 'sub_location'] = extracted.combine_first(df.loc[extracted.index, 'sub_location'])
```

```python
import re

PROJECT_PATTERNS = [
    ('Sawera Mega Residency', r'sawera mega residency'),
    ('Sawera Mega', r'sawera mega(?!\s+residency)'),
    ('Sawera Icon', r'sawera icon'),
    ('Chapal Navinta', r'chapal navinta'),
    ('Taj Complex', r'taj complex'),
    ('Avari Towers', r'avari towers?'),
    ('Empress Market', r'empress market'),
    ('Karachi Press Club', r'karachi press club'),
    ('Saddar Town', r'saddar town'),
]

def extract_sub_location(text):
    if not isinstance(text, str):
        return None
    for name, pat in PROJECT_PATTERNS:
        if re.search(pat, text, re.IGNORECASE):
            return name
    return None


extracted = x.apply(extract_sub_location)


df.loc[extracted.index, 'sub_location'] = extracted.combine_first(df.loc[extracted.index, 'sub_location'])
```

```python
import pandas as pd
import numpy as np


MIN_COUNT = 10


K = 10


location_stats = (
    df.groupby("location")
      .agg(
          location_median_ppsqft=("price_per_sqft", "median"),
          location_count=("price_per_sqft", "count")
      )
      .reset_index()
)


sub_stats = (
    df.groupby(["location", "sub_location"])
      .agg(
          sub_median_ppsqft=("price_per_sqft", "median"),
          sub_count=("price_per_sqft", "count")
      )
      .reset_index()
)




sub_stats = sub_stats.merge(
    location_stats,
    on="location",
    how="left"
)

sub_stats["sub_weight"] = (
    sub_stats["sub_count"] /
    (sub_stats["sub_count"] + K)
)

sub_stats["parent_weight"] = (
    K /
    (sub_stats["sub_count"] + K)
)


sub_stats["adjusted_ppsqft"] = (
    sub_stats["sub_weight"] *
    sub_stats["sub_median_ppsqft"]
    +
    sub_stats["parent_weight"] *
    sub_stats["location_median_ppsqft"]
)


sub_stats["adjusted_ppsqft"] = (
    sub_stats["adjusted_ppsqft"]
    .fillna(sub_stats["location_median_ppsqft"])
)

sub_stats["price_percentile"] = (
    sub_stats["adjusted_ppsqft"]
    .rank(
        method="average",
        pct=True
    )
)



# Turn each adjusted percentile into a score from 1 to 5.


sub_stats["sub_location_score"] = pd.cut(
    sub_stats["price_percentile"],
    bins=[
        0,
        0.20,
        0.40,
        0.60,
        0.80,
        1.00
    ],
    labels=[
        1,
        2,
        3,
        4,
        5
    ],
    include_lowest=True
).astype(int)


# Build a lookup table for each location and sub-location pair.

score_map = (
    sub_stats
    .set_index(["location", "sub_location"])
    ["sub_location_score"]
)


# Add the calculated sub-location score back to the main dataframe.

df["sub_location_score"] = [
    score_map.get((loc, sub), np.nan)
    for loc, sub in zip(
        df["location"],
        df["sub_location"]
    )
]



# Use the parent location's score for rows with an unspecified sub-location.
#
# For example, DHA Defence_Unspecified does not identify a specific area,
# so it receives the broader DHA Defence market score.


unknown_mask = (
    df["sub_location"]
    .astype("string")
    .str.endswith("_Unspecified", na=False)
)


# Rank parent locations by their median price per square foot.
location_percentile_map = (
    location_stats
    .set_index("location")
    ["location_median_ppsqft"]
    .rank(pct=True)
)


location_score_map = pd.cut(
    location_percentile_map,
    bins=[
        0,
        0.20,
        0.40,
        0.60,
        0.80,
        1.00
    ],
    labels=[
        1,
        2,
        3,
        4,
        5
    ],
    include_lowest=True
).astype(int)


df.loc[unknown_mask, "sub_location_score"] = (
    df.loc[unknown_mask, "location"]
      .map(location_score_map)
)


# Fill scores for records where the sub-location itself is missing.

missing_sub_mask = df["sub_location"].isna()

df.loc[missing_sub_mask, "sub_location_score"] = (
    df.loc[missing_sub_mask, "location"]
      .map(location_score_map)
)


# Use the parent-location score as the final fallback for any remaining gaps.

df["sub_location_score"] = (
    df["sub_location_score"]
    .fillna(
        df["location"]
        .map(location_score_map)
    )
)


# Round the scores and keep them within the expected 1-to-5 range.

df["sub_location_score"] = (
    df["sub_location_score"]
    .round()
    .clip(1, 5)
    .astype(int)
)


# Print a sample of the results so the scoring can be checked.

print(
    sub_stats[
        [
            "location",
            "sub_location",
            "sub_count",
            "sub_median_ppsqft",
            "location_median_ppsqft",
            "sub_weight",
            "parent_weight",
            "adjusted_ppsqft",
            "sub_location_score"
        ]
    ]
    .sort_values(
        "adjusted_ppsqft",
        ascending=False
    )
    .head(30)
)


# Show how many records received each score.

print(
    "\nSub-location score distribution:"
)

print(
    df["sub_location_score"]
    .value_counts()
    .sort_index()
)
```
