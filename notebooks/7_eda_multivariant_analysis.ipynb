```python
#remaning columns

hue_cols = ['sub_location_score', 'bathrooms', 'parking_score',
            'servant_quarter_score', 'electricity_backup',
            'furnished', 'sub_location_type']

df_filtered = df[df['area'] < 10000]

n_cols = 3
n_rows = -(-len(hue_cols) // n_cols)
fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, n_rows * 4))
axes = axes.flatten()


for i, col in enumerate(hue_cols):
    sns.scatterplot(data=df_filtered,
        x='area', y=df_filtered['price'] / 10_000_000,
        hue=col,
        alpha=0.7,
        ax=axes[i]
    )

    axes[i].set(title=f'Area vs Price by {col}', xlabel='Area (sqft)', ylabel='Price (Crore PKR)')
    axes[i].legend(bbox_to_anchor=(1, 1), fontsize=8)


for ax in axes[len(hue_cols):]:
    fig.delaxes(ax)

plt.tight_layout()
plt.show()
```

#Observation
* Higher bathroom counts (5+) and active servant quarters concentrate tightly within the top price tiers across larger square footage categories, showing a strong link to luxury builds.

* Properties featuring dedicated parking spaces (score $\ge 1$) hold a structurally higher price floor across every size bracket compared to non-parking counterparts.

* Both electricity backup availability and furnishing status spread evenly across all pricing brackets, demonstrating that they are baseline utility expectations rather than features commanding price premiums.

* Housing societies and military housing sub-location types drive the most aggressive price surges ($>40$ Crore PKR) on large-scale standard plot areas, while regular residential projects flatten out at lower valuations.

* Sub-location scores act as an effective vertical price gradient; darker, highly-rated locations systematically pull property values toward the ceiling of each individual area cluster.
